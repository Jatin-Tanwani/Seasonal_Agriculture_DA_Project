```python
# Fill missing Rainfall using the median of the corresponding Season
rainfall_medians = df.groupby('Season')['Rainfall_mm'].median()
df['Rainfall_mm'] = df['Rainfall_mm'].fillna(
    df.groupby('Season')['Rainfall_mm'].transform('median')
)
print("Missing Rainfall values before:", 48)
print("Missing Rainfall values after:", df['Rainfall_mm'].isnull().sum())
print("\nSeason-wise medians used:")
print(rainfall_medians)
```

```python

soil_medians = df.groupby('Season')['Soil_Moisture_pct'].median()
df['Soil_Moisture_pct'] = df['Soil_Moisture_pct'].fillna(
    df.groupby('Season')['Soil_Moisture_pct'].transform('median')
)
print("Missing Soil Moisture values before:", 40)
print("Missing Soil Moisture values after:", df['Soil_Moisture_pct'].isnull().sum())
print("\nSeason-wise medians used:")
print(soil_medians)
```

```python

rows_before = len(df)
df = df.dropna(subset=['Yield_Tonnes_Ha']).copy()
rows_after = len(df)

print("Rows before removing missing Yield:", rows_before)
print("Rows removed:", rows_before - rows_after)
print("Rows after cleaning Yield:", rows_after)
```

```python

text_columns = df.select_dtypes(include='object').columns
for col in text_columns:
    df[col] = df[col].str.strip()

print("Text columns standardized:", list(text_columns))
```
